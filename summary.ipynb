```python
from langchain.prompts import PromptTemplate
from langchain.chains.llm import LLMChain
from langchain_openai import ChatOpenAI

```

```python
llm = ChatOpenAI(model="gpt-4o")

prompt = PromptTemplate(
    input_variables=["text"],
    template="Summarize this: {text}"
)
```

```python
chain = LLMChain(llm=llm, prompt=prompt)
```

```python
print(chain.run("The efficiency of single-modality techniques is severely limited by the minute changes in the structure and color of the retina, which only partially reflect on the images, enabling the diagnosis of retinal illnesses difficult. In order to do this, this work proposes an AI-driven multimodal framework that independently evaluates fundus and Optical Coherence Tomography (OCT) images for the diagnosis of retinal diseases and personalized treatment recommendations. Separate deep learning pipelines are developed to maintain modality-specific features since fundus images are RGB and texture-rich, whereas OCT images are monochrome and structure-focused Different CNN models have been reviewed, such as ResNet-50, ResNet-101, RegNetY-8GF, RegNetY-16GF, and several EfficientNet variations. The comparison results show that EfficientNet-B3 is the best for fundus analysis; however, for the OCT classification EfficientNet-B4 is the best choice, since the two models retain high accuracy and computational efficiency. The OCT technique can accurately classify eight different retinal disease types with a 96% success rate on the test set, whereas the fundus method obtains a classification accuracy of 88% for seven retinal conditions. To make the models more understandable and gain the clinicians' trust, Grad CAM++ has been combined with both models to visualize the disease and the particular retinal areas that were influential in the prediction. Additionally, to provide personalized clinical advice, a Random Forest-based recommendation module makes use of patient demographics, disease severity indicators, explainability-derived insights, and model confidence scores. The suggested framework exhibits clinically significant recommendations, interpretable decision-making, as well as diagnosis. This demonstrates its potential as a useful tool for ophthalmologists for applications involving complete examinations for retinal diseases.."))
```
